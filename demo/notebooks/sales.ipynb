# Q3 sales review

Revenue by region, plus a quick Q4 forecast.

```python
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
sales = pd.DataFrame({
    'region': ['North', 'South', 'East', 'West'],
    'revenue': [120, 95, 140, 80],
})
sales
```

```python
growth = 1.10
forecast = sales.assign(q4=sales.revenue * growth)
forecast
```

```python
forecast.plot.bar(x='region', y=['revenue', 'q4'])
plt.title('Revenue: Q3 vs Q4 forecast')
plt.show()
```
